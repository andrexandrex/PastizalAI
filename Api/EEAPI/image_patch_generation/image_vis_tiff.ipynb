```python
import re
import rasterio
import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np

from shapely.geometry import box, Point
from rasterio.plot import show
from rasterio.windows import Window


def plot_sar_patches(
    image_path,
    shapefile_path,
    spills_df,
    patch_size=512,
    n=3,  # number of patches “along” the perpendicular line
    m=1,  # number of “rows” above/below that line
):
    """
    Plots a Sentinel-1 image, overlays the region boundary and the spill point,
    then computes and plots a series of 512x512 patches along a perpendicular
    (tangent to the boundary) from the sea to the spill point (and optionally
    above/below).

    Parameters:
    -----------
    image_path : str
        Full path to the Sentinel-1 GeoTIFF image.
    shapefile_path : str
        Full path to the DEPARTAMENTOS.shp file.
    spills_df : pd.DataFrame or gpd.GeoDataFrame
        Spill data with columns 'LAT', 'LON', 'REGION', 'Original index'.
    patch_size : int
        Size (in pixels) for each square patch (default 512).
    n : int
        How many patches to take along the line from ocean to the spill point.
    m : int
        How many “rows” above/below the perpendicular line to include.

    Returns:
    --------
    None (displays two matplotlib plots):
        1) The overall image with boundary and spill point.
        2) The patches, each 512×512, marked on the same figure.
    """

    # --- Parse index from filename ---
    image_filename = image_path.split("/")[-1].split("\\")[-1]
    match = re.match(r"(\d+)_", image_filename)
    if not match:
        raise ValueError("Could not parse index from filename.")
    image_index = int(match.group(1))
    print(f"[INFO] Parsed index from image: {image_index}")

    # --- Convert spills_df to GeoDataFrame if needed ---
    if not isinstance(spills_df, gpd.GeoDataFrame):
        spills_df["geometry"] = spills_df.apply(
            lambda row: Point(row["LAT"], row["LON"]), axis=1
        )
        spills_gdf = gpd.GeoDataFrame(spills_df, geometry="geometry", crs="EPSG:4326")
    else:
        spills_gdf = spills_df

    # --- Filter spill points for this image_index ---
    filtered_spills = spills_gdf[spills_gdf["Original index"] == image_index]
    if len(filtered_spills) == 0:
        raise ValueError(f"[ERROR] No spill found for index {image_index}!")
    # We'll just take the first row if more exist:
    spill_row = filtered_spills.iloc[0]

    # --- Read SAR image ---
    with rasterio.open(image_path) as src:
        img = src.read(1)
        crs = src.crs
        transform = src.transform
        bounds = src.bounds
        image_extent = box(*bounds)

    # --- Load shapefile and filter to region ---
    peru_gdf = gpd.read_file(shapefile_path)
    if peru_gdf.crs is None:
        peru_gdf.set_crs(epsg=4326, inplace=True)

    # region name from spill data
    target_region = spill_row["REGION"]
    # Filter the departmental shapefile
    region_gdf = peru_gdf[peru_gdf["NOMBDEP"] == target_region]

    if region_gdf.empty:
        raise ValueError(f"[ERROR] No matching region in shapefile for '{target_region}'")

    # Reproject to match the SAR image
    if peru_gdf.crs != crs:
        region_gdf = region_gdf.to_crs(crs)
        filtered_spills = filtered_spills.to_crs(crs)

    # We'll assume region_gdf is a single polygon (or we take the first)
    region_polygon = region_gdf.geometry.iloc[0]
    # If it's multi-part, you might need to pick the part nearest the spill, etc.
    region_boundary = region_polygon.exterior

    # Our spill point in the region's CRS
    spill_geom = filtered_spills.geometry.iloc[0]  # just one
    if not spill_geom.is_valid:
        spill_geom = spill_geom.buffer(0)

    # --- Find the boundary coordinate that is closest to the spill point ---
    # param along boundary:
    param = region_boundary.project(spill_geom)
    coast_closest = region_boundary.interpolate(param)

    # Approximate the local tangent of boundary near that param
    EPS = 1e-6
    tangent_pt2 = region_boundary.interpolate(param + EPS)
    # Vector from coast_closest -> tangent_pt2
    dx = tangent_pt2.x - coast_closest.x
    dy = tangent_pt2.y - coast_closest.y
    tangent_vec = np.array([dx, dy])
    if np.allclose(tangent_vec, 0):
        # fallback if boundary is weird or param is at an extreme
        tangent_vec = np.array([1.0, 0.0])

    # Normalize tangent
    tangent_len = np.hypot(*tangent_vec)
    tangent_dir = tangent_vec / tangent_len

    # A perpendicular to tangent: rotate 90 deg
    # ( x, y ) -> ( -y, x )
    normal_dir = np.array([-tangent_dir[1], tangent_dir[0]])

    # We want to decide which side is “sea”.  As a simple guess, we check
    # whether moving along +normal or -normal is closer/further from the region polygon.
    # This is very heuristic.  You could refine for real cases!
    def dist_to_poly(pt):
        return region_polygon.exterior.distance(Point(pt))

    # Evaluate distance from a point a small step away in +normal
    test_plus = spill_geom.coords[0] + normal_dir * 1000  # 1 km step, for instance
    test_minus = spill_geom.coords[0] - normal_dir * 1000

    dist_plus = dist_to_poly(test_plus)
    dist_minus = dist_to_poly(test_minus)

    # If plus is bigger, that means plus is “outside” (sea). Otherwise, minus is sea.
    sea_side_dir = normal_dir if dist_plus > dist_minus else -normal_dir

    # --- Convert patch size (pixels) to actual ground distance. ---
    # You can approximate by the average resolution from the transform.
    # E.g. if the image is in meters
    res_x = transform.a  # pixel width
    res_y = transform.e  # typically negative for north-up images
    # We'll take absolute values for resolution:
    mean_res = abs(res_x + res_y) / 2  # crude average
    patch_ground_size = patch_size * mean_res

    # We'll create a grid of patch “centers” along sea_side_dir, from offshore (n patches away)
    # up to the actual spill point, and for “m” rows above and below that line.
    # For example: i=0 => the spill location, i=1 => one patch_size away from spill, etc.

    patch_centers = []
    spill_x, spill_y = spill_geom.x, spill_geom.y

    # how many patches from the sea side to the point
    for i in range(n + 1):
        # offset along the normal
        offset = (n - i) * patch_ground_size  # i=0 => offset=n * ...
        center_base = np.array([spill_x, spill_y]) + sea_side_dir * offset

        # now add the top/bottom rows, which effectively means moving along
        # the “tangent_dir” or some perpendicular to sea_side_dir
        # ( sea_side_dir is normal; tangent_dir is tangent).
        for j in range(-m, m + 1):
            # shift in the tangent direction
            center_shift = tangent_dir * j * patch_ground_size
            center = center_base + center_shift
            patch_centers.append(center)

    # --- Build bounding boxes for each patch in pixel-coordinates, then read them ---
    patch_windows = []
    with rasterio.open(image_path) as src:
        for cx, cy in patch_centers:
            # transform to pixel coords
            px, py = ~src.transform * (cx, cy)
            # top-left corner
            row_off = int(py - patch_size / 2)
            col_off = int(px - patch_size / 2)
            # Construct a rasterio Window
            win = Window(col_off=col_off, row_off=row_off, width=patch_size, height=patch_size)

            # If you want to ensure the window is fully inside the image, you might need checks:
            if (
                row_off < 0
                or col_off < 0
                or (row_off + patch_size) > src.height
                or (col_off + patch_size) > src.width
            ):
                # skip patches that go out of bounds
                continue

            patch_windows.append(win)

    # --- Plot everything ---
    # 1) Plot the base image, boundary, spill point:
    fig, ax = plt.subplots(figsize=(10, 10))
    with rasterio.open(image_path) as src:
        show((src, 1), ax=ax, cmap="gray")
    region_gdf.boundary.plot(ax=ax, edgecolor="red", linewidth=1.5, label="Region boundary")
    filtered_spills.plot(ax=ax, color="cyan", markersize=30, label="Spill point", zorder=5)
    ax.set_xlim(bounds.left, bounds.right)
    ax.set_ylim(bounds.bottom, bounds.top)
    ax.set_title(f"SAR Image + Boundary + Spill (Index: {image_index})")
    ax.legend()
    plt.tight_layout()

    # 2) Plot the same figure but show patch rectangles
    # (We’ll overlay rectangles to indicate where we’ll extract the 512×512 patches)
    for win in patch_windows:
        # Convert window corners back to map coords to plot them as a rectangle
        left_px, top_px = win.col_off, win.row_off
        right_px, bottom_px = left_px + win.width, top_px + win.height

        # Map coords
        top_left = src.transform * (left_px, top_px)
        top_right = src.transform * (right_px, top_px)
        bot_left = src.transform * (left_px, bottom_px)
        bot_right = src.transform * (right_px, bottom_px)

        # Build a polygon
        patch_poly = box(
            min(top_left[0], top_right[0], bot_left[0], bot_right[0]),
            min(top_left[1], top_right[1], bot_left[1], bot_right[1]),
            max(top_left[0], top_right[0], bot_left[0], bot_right[0]),
            max(top_left[1], top_right[1], bot_left[1], bot_right[1]),
        )
        x_patch, y_patch = patch_poly.exterior.xy
        ax.plot(x_patch, y_patch, color="yellow", linewidth=1.5)

    plt.show()

    # If you want to actually extract the patch data from the raster for
    # further analysis, you can do something like:
    # with rasterio.open(image_path) as src:
    #     for i, win in enumerate(patch_windows):
    #         patch_data = src.read(1, window=win)
    #         # do something with patch_data (classification, saving, etc.)
```
